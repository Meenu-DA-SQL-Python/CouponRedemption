```python
import pandas as pd
from datasets import load_dataset
import numpy as np
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])


df.info()
```

```python
job_count_title=df['job_title_short'].value_counts().sort_values(ascending=True)
```

```python
plt.bar(job_count_title.index,job_count_title)
plt.title('Postngs by Job title')
plt.ylabel('Count of job postings')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Problem Statement:
# Using the same bar chart created in Matplotlib Plotting Question #1 showing the number of job postings for the job_country:

# Use the job_country column and count the number of postings for each location. Only get the top 5 job countries.
# Add labels for the x-axis (label as 'Country'), y-axis (label as 'Number of Postings'), 
# and a title (set as 'Number of Jobs per Country') for the chart.
# Rotate the x-axis labels by 45 degrees and align them to the right.


posting_by_country=df['job_country'].value_counts().sort_values(ascending=False)
print(posting_by_country)

posting_by_country=posting_by_country.head(5)
print(posting_by_country)

plt.bar(posting_by_country.index,posting_by_country,color='yellow')
plt.xlabel='Country'
plt.ylabel='Number of Postings'
plt.xticks(rotation=45,ha='right')
```

```python
# Problem Statement:
# Using the same bar chart created in Matplotlib Plotting Question #2 showing the average yearly salary for different job titles:

# Use the 'job_title_short' and salary_year_avg columns to calculate the average salary for each job title.
# Add labels for the x-axis (label as 'Average Salary') and a title (set as 'Average Salary by Job Title') for the plot.


jobtitle_avgsalary=df.groupby('job_title_short')['salary_year_avg'].mean().sort_values()
print(jobtitle_avgsalary)

plt.barh(jobtitle_avgsalary.index,jobtitle_avgsalary,color='purple')
plt.title('Average Salary by Job Title')
plt.xlabel ='Average Salary'
plt.ylabel ='Average Salary by Job Title'
```

```python
# Problem Statement:
# Using the same line plot created in Matplotlib Plotting Question #3 showing the trend of average yearly salary over
#the first 6 months of the year:

# Use the job_posted_date column to extract the month and calculate the average salary for each month.
# Add labels for the x-axis (label as 'Month'), y-axis (label as 'Average Salary'), and a title (set as 'Average Salary by Job Month') 
#for the plot.
# Change the line color to 'red' and add a legend to the plot.

df['job_posted_month']=df['job_posted_date'].dt.month




jobmonth_avgsalary=df.groupby('job_posted_month')['salary_year_avg'].mean().sort_index()
print(jobmonth_avgsalary)

plt.plot(jobmonth_avgsalary.index,jobmonth_avgsalary,color='red',label='Average Salary by Mnt')
plt.title('Average Salary by Month Posted')
plt.xlabel='month'
plt.xlabel ='Average Salary'
plt.ylabel ='Average Salary by Month posted'
plt.legend()
plt.show()
```
