Classes: The Template of Objects
Class:A template for creating objects

```python
class jobpost:
    def __init__(self,title,location,salary):
        self.title=title
        self.location=location
        self.salary=salary
```

```python
jobpost(job_title,job_location,total_salary)
```

Methods:the functions of an object
method; A Function defined inside a class that operates on its objects

```python
class jobpost:
    def __init__(self,title,location,salary):
        self.title=title
        self.location=location
        self.salary=salary
    def aboutjob(self):
        return print(f" Job Title  :  {job_title}\n Location:  {job_location}\nSalary:  {total_salary}")
```

```python
job1=jobpost(job_title,job_location,total_salary)
```

```python
job1.title
```

```python
job1.aboutjob()
```
