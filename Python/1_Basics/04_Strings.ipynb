<span style = "color:blue ; font-weight :bold ; font-size :32px; "> __format__</span>  
4.__format__(self, format_spec)

```python
name='Mullai'
width=15
#1.1 
print('Hello' +' '+ 'Meenu'.ljust(10))
#1.2
print('Hello' +' '+ 'Meenu'.center(10))
#1.3
print('Hello' +' '+ 'Meenu'.rjust(10))
#1.4
print('Hello'+' '+ 'Meenu'.ljust(10,'*'))
#1.5
print('Hello'+' '+ name.center(10,'$'))
#1.6
print('Hello'+' '+ name.rjust(10,'$'))
#1.7 Dynamic Width
print('Hello'+' '+name.center(width,'@'))

print('Hello' +' '+ '{:*^10}'.format('Meenu'))
```

<span style = "color:blue ; font-weight :bold ; font-size :32px; "> __rmod__</span>  
18. __rmod__(self, value) → value % self

```python
class ModTest:
    def __mod__(self, other):
        return "Left-side __mod__ called"

    def __rmod__(self, other):
        return "Right-side __rmod__ called"

obj = ModTest()

print(obj % 10)  # Calls obj.__mod__(10)
print(10 % obj)  # Calls obj.__rmod__(10)
```
