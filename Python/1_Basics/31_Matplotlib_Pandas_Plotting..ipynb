```python
import pandas as pd
from datasets import load_dataset
import numpy as np
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])


df.info()
```

```python
job_count_title=df['job_title_short'].value_counts().sort_values(ascending=False)

plt.bar(job_count_title.index,job_count_title)

plt.title('Postngs by Job title')
plt.ylabel('Count of job postings')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
job_count_title.plot(kind='bar')
plt.title('Postngs by Job title')
plt.xlabel('')
plt.ylabel('Count of job postings')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
df = df.sort_values(by='job_posted_date') 
# Sort DataFrame by date
df.plot(x='job_posted_date', y='salary_year_avg', kind='line')  # Now plot the sorted data


```

```python
# Problem Statement:
# Create a bar plot with the Pandas .plot() method showing the top 5 job posting platforms 
# (e.g., LinkedIn, Upwork, etc.) with the highest number of job postings. 
# Use the job_via column to count the number of postings for each platform.

job_via_count=df['job_via'].value_counts().sort_values(ascending=False)

job_via_count=job_via_count.head(5)

print(job_via_count)

job_via_count.plot(kind='bar')
```

```python
# Problem Statement:
# Create a line plot showing the trend of job postings over time using the Pandas method to plot:
# Create a DataFrame called monthly_job_counts and use the job_posted_date column to count the number of job postings for each month 
# (in the format 'Jan').
# Add labels for the x-axis (label as 'Month'), y-axis (label as 'Number of Job Postings'), 
# and a title (set as 'Trend of Job Postings Over Time') for the chart.

```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_posted_month'] = df['job_posted_date'].dt.to_period('M')

monthly_job_counts = df['job_posted_month'].value_counts().sort_index()

monthly_job_counts.plot(kind='line')
```

```python
# Problem Statement:
# Create a horizontal bar plot showing the top 5 job locations with the highest number of job postings:

# Use the job_location column to count the number of postings for each location.
# Add in labels for the x-axis (label as 'Count of Jobs'), y-axis (label as 'Job Location'), 
# and a title (set as 'Number of Job Postings by Job Location') for the chart.

location_counts = df['job_location'].value_counts().head(5)
location_counts.plot(kind='barh')
plt.xlabel('Count of Jobs')
plt.ylabel('Job Location')
plt.title('Number of Job Postings by Job Location')
plt.show()

```
