### Notes 

`if`, `elif`, and `else` are conditional statements that control the flow of execution based on whether a condition is `True` or `False`.

| Part       | Purpose                           | Condition Evaluated             | Action if True                 |
|------------|-----------------------------------|---------------------------------|--------------------------------|
| `if`       | First condition check             | First condition to be evaluated | Executes block if condition is true |
| `elif`     | Subsequent condition check        | Next condition, if `if` is false| Executes block if this condition is true |
| `else`     | Fallback if no conditions are met | None | Executes block if all above conditions are false |


```python
if True:
    print('You are good in Christ')
```

```python
job_skill='SQL'
applicant_skill='SQL'
if job_skill==applicant_skill:
    print('Skill Match')
```

```python
# Determine the job_title based on years of job experience and years of SQL experience. The variable job_experience is set to 4 and sql_experience is set to 2. Use the following criteria:

## Less than 2 years of job experience:
# SQL experience → "Associate Analyst"
# 1+ years of SQL experience → "Junior Analyst"

## 2 to 5 years of job experience:
# 0 SQL experience → "Associate Analyst"
# 1 year of SQL experience → "Intermediate Analyst"
# 2+ years of SQL experience → "Senior Analyst"

## More than 5 years of job experience:
# 0–2 years of SQL experience → "Data Analyst"
# 3+ years of SQL experience → "Lead Analyst"

job_experience = 8
sql_experience = 7

if job_experience<2:
    if sql_experience==0:
        job_title="Associate Analyst"
    else:
        job_title="Junior Analyst"  
elif job_experience<5:
    if sql_experience==0:
        job_title="Associate Analyst"
    elif sql_experience<1:
        job_title="Intermediate Analyst"
    else :
        job_title="Senior Analyst"  
else: # 2 to 5 years of experience
    if sql_experience == 0:
        job_title = "Associate Analyst"
    elif sql_experience == 1:
        job_title = "Intermediate Analyst"
    else:
        job_title = "Senior Analyst"

print(job_title)

```
