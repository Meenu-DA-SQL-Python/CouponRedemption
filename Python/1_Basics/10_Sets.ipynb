## Notes

* For unique items
* Used to store multiple items in a single variable
* It is:
  * unorderd
  * Unchangeable
  * Unindexed
* Set *items* are unchangeable but you can still remove items and add new items

## Importance

Employed for removing duplicates and for set operations, aiding in data cleaning and preparation.

## Examples

First let's create a set of data science skills.

```python
restaurants={5,'food_spicelevel','Non_Veg','5* Rating'}

restaurants
```

```python
help(set)
```

```python
restaurants.add(5)
restaurants.add('5*Rating')
print(restaurants)
```

```python
restaurants.pop() #Random removing
print(restaurants)
```

```python
restaurants.add('Seating')
print(restaurants)
```

```python
restaurants.remove(5)
```

```python
best_food=['briyani','mutton curry','payasam','Idly','dosa','briyani','Sambar','Sambar','Idly']
best_food=list(set(best_food))
print(best_food)
```

```python
# Problem Statement:
# Add a new job title 'AI Specialist' to the set of unique job titles and print the updated set. 
# The initial set of unique_job_titles is {'Data Scientist', 'Data Analyst', 'Machine Learning Engineer'}.

unique_job_titles = {'Data Scientist', 'Data Analyst', 'Machine Learning Engineer'}

unique_job_titles.add('AI Specialist')

print(unique_job_titles)
```

```python
# Problem Statement:
# Remove the job title 'Data Analyst' from the set of unique job titles and print the updated set. 
# The initial set of unique_job_titles is {'Data Scientist', 'Data Analyst', 'Machine Learning Engineer'}.

unique_job_titles = {'Data Scientist', 'Data Analyst', 'Machine Learning Engineer'}

unique_job_titles.remove('Data Analyst')

print(unique_job_titles)
```

```python
#Problem Statement:
#Create a set of job locations named job_locations from a list of job locations for data science roles and print the set.


job_locations = ['New York', 'San Francisco', 'New York', 'Austin', 'San Francisco']

job_locations = set(job_locations)

print(job_locations)
```

```python
#Problem Statement:
#Find the union of two sets of job skills and print the result.


skills_set1 = {'Python', 'SQL', 'Tableau'}
skills_set2 = {'R', 'SQL', 'Machine Learning'}
all_skills = skills_set1.union(skills_set2)
print(all_skills)
```
