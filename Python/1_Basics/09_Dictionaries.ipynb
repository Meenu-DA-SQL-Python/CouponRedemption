# 📌 Python Dictionary Methods with Examples

## 1️⃣ Accessing Data
| **Method** | **Description** | **Example** |
|------------|----------------|-------------|
| `dict.get(key, default)` | Returns the value for the key; if not found, returns default. | `job_postings.get('Data Scientist', 'Not Found')` → `120` |
| `dict.keys()` | Returns all keys in the dictionary. | `job_postings.keys()` → `dict_keys(['Data Scientist', 'Data Analyst'])` |
| `dict.values()` | Returns all values in the dictionary. | `job_postings.values()` → `dict_values([120, 80])` |
| `dict.items()` | Returns key-value pairs as tuples. | `job_postings.items()` → `[('Data Scientist', 120), ('Data Analyst', 80)]` |

---

## 2️⃣ Updating & Modifying Data
| **Method** | **Description** | **Example** |
|------------|----------------|-------------|
| `dict.update(another_dict)` | Merges another dictionary into the existing one. | `job_postings.update({'ML Engineer': 50})` |
| `dict.setdefault(key, default)` | Returns value if key exists; otherwise, sets it to the default. | `job_postings.setdefault('AI Specialist', 30)` |

---

## 3️⃣ Removing Data
| **Method** | **Description** | **Example** |
|------------|----------------|-------------|
| `dict.pop(key, default)` | Removes the key and returns its value; default if not found. | `job_postings.pop('Data Scientist')` → `120` |
| `dict.popitem()` | Removes and returns the last key-value pair. | `job_postings.popitem()` → `('Data Analyst', 80)` |
| `dict.clear()` | Removes all elements from the dictionary. | `job_postings.clear()` → `{}` |

---

## 4️⃣ Copying & Creating Dictionaries
| **Method** | **Description** | **Example** |
|------------|----------------|-------------|
| `dict.copy()` | Creates a shallow copy of the dictionary. | `new_dict = job_postings.copy()` |
| `dict.fromkeys(keys, value)` | Creates a new dictionary with keys and the same value. | `dict.fromkeys(['a', 'b'], 0)` → `{'a': 0, 'b': 0}` |

---

## 5️⃣ Other Methods
| **Method** | **Description** | **Example** |
|------------|----------------|-------------|
| `dict.__len__()` | Returns the number of items in the dictionary. | `len(job_postings)` → `2` |
| `dict.__contains__(key)` | Checks if a key exists in the dictionary. | `'Data Scientist' in job_postings` → `True` |
| `dict.__delitem__(key)` | Deletes a key-value pair from the dictionary. | `del job_postings['Data Scientist']` |
| `dict.__eq__(other_dict)` | Compares two dictionaries for equality. | `job_postings == {'Data Scientist': 120}` → `True` |

---

````python
# Problem Statement:
# Create a nested dictionary job_details where each key is a job role and the value is another dictionary containing 'postings' and 
# 'average_salary'. 
# The details for each are:
#     'Data Scientist' has 120 postings and an average salary of 120000; 'Data Analyst' has 80 postings and an average salary of 80000; 
# Machine Learning Engineer has 50 postings and average salary of 110000. 
#Print the nested dictionary.

job_details = {
    'job role': {
        'Data Scientist': (120, 120000),
        'Data Analyst': (80, 80000),
        'Machine Learning Engineer': (50, 110000)
    }
}


role='Data Scientist'

postings,salary=job_details['job role'][role]

print(f" {role} has {postings} postings and average salary of {salary}.")

print(job_details.get('Data','Notfound'))

print(job_details['job role'].get('Data Analyst','Not Found'))

#print(job_details['job role']['Data Analyst'].items())--this won't work as .items is only for dictionary

job_details['job role'].update({'Business Analyst':(10,150000)})

print(job_details)

job_details.setdefault('Design Engineer', 'CADD')

job_details.popitem()

print(job_details)

job_details_copy=job_details.copy()

print(job_details_copy)
````
